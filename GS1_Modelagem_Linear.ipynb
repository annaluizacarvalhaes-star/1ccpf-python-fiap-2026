EX 1 - Gere dois gráficos distintos

```python
# Qualidade de período
import pandas as pd
import matplotlib.pyplot as plt

dados = {
    'nome': ['Ryugu', '1989 ML', 'Nereus', 'Bennu', 'Didymos',
             '2011 UW158', 'Anteros', '2001 CC21', '1992 TC', '2001 SG10'],
    'qualidade_periodo': [2, 3, 3, 3, 3, 3, 3, 3, 2, 0],
    'velocidade': [4.663, 4.889, 4.987, 5.096, 5.162,
                   5.189, 5.440, 5.636, 5.648, 5.880]
}

df = pd.DataFrame(dados)

plt.figure(figsize=(10,5))
plt.bar(df['nome'], df['qualidade_periodo'])
plt.title('Qualidade do Período dos Asteroides')
plt.xlabel('Asteroides')
plt.ylabel('Qualidade do Período')
plt.xticks(rotation=45)
plt.show()
```

```python
# Qualidade de período x Velocidade
plt.figure(figsize=(8,5))
plt.scatter(df['velocidade'], df['qualidade_periodo'])

plt.title('Velocidade x Qualidade do Período')
plt.xlabel('Velocidade (km/s)')
plt.ylabel('Qualidade do Período')

plt.show()
```

```python
import pandas as pd

dados = {
    'nome': ['Ryugu', '1989 ML', 'Nereus', 'Bennu', 'Didymos',
             '2011 UW158', 'Anteros', '2001 CC21', '1992 TC', '2001 SG10'],
    'qualidade_periodo': [2, 3, 3, 3, 3, 3, 3, 3, 2, 0],
    'velocidade': [4.663, 4.889, 4.987, 5.096, 5.162,
                   5.189, 5.440, 5.636, 5.648, 5.880]
}

df = pd.DataFrame(dados)
```

```python
# 1. análise qualidade do período
coluna = df['qualidade_periodo']

print("Média:", coluna.mean())
print("Mediana:", coluna.median())
print("Moda:", coluna.mode()[0])

print("Máximo:", coluna.max())
print("Mínimo:", coluna.min())
print("Amplitude:", coluna.max() - coluna.min())
print("Variância:", coluna.var())
print("Desvio padrão:", coluna.std())

print("Q1:", coluna.quantile(0.25))
print("Q2 (Mediana):", coluna.quantile(0.50))
print("Q3:", coluna.quantile(0.75))
```

```python
# 2. Análise Velocidade (km/s)
coluna = df['velocidade']

print("Média:", coluna.mean())
print("Mediana:", coluna.median())
print("Moda:")
print(coluna.mode())

print("Máximo:", coluna.max())
print("Mínimo:", coluna.min())
print("Amplitude:", coluna.max() - coluna.min())
print("Variância:", coluna.var())
print("Desvio padrão:", coluna.std())

print("Q1:", coluna.quantile(0.25))
print("Q2 (Mediana):", coluna.quantile(0.50))
print("Q3:", coluna.quantile(0.75))
```
